```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.preprocessing import LabelEncoder
```

<b>Task:</b> Use the rest of this notebook to carry out your project plan. 

You will:

Add code cells below and populate the notebook with commentary, code, analyses, results, and figures as you see fit. 

```python
# load the dataset
print("Airbnb Dataset loading...")
df = pd.read_csv(airbnbDataSet_filename)
print(f"Dataset loaded: {df.shape[0]} rows, {df.shape[1]} columns")


print("STEP 1: DATA PREPARATION")
print("Cleaning price data...")
print(f"Price column info before cleaning:")
print(f"Data type: {df['price'].dtype}")
print(f"Sample values: {df['price'].head()}")
      
# cleaning the price column by removing the dollar signs and commas--converting them numerically
df['price'] = df['price'].astype(str).str.replace('$', '').str.replace(',', '')
df['price'] = pd.to_numeric(df['price'], errors='coerce')
print(f"\nPrice statistics after cleaning:")
print(df['price'].describe())
      
      
# removing outliers that are greater than $10 or less than $500
print(f"\nRemoving outliers...")
print(f"Original dataset size: {len(df)}")
df = df[(df['price'] >= 10) & (df['price'] <= 500)]
print(f"After outlier removal: {len(df)} rows")
      
# selecting specific features that I mentioned in my plan 
print(f"\nDecided features...")
feature_columns = [
    'host_is_superhost',
    'instant_bookable', 
    'review_scores_rating',
    'accommodates',
    'bedrooms',
    'bathrooms',
    'neighbourhood_group'
]

# identify which columns exist in the dataset
available_features = []
for col in feature_columns:
    if col in df.columns:
        available_features.append(col)
        print(f"✓ {col} - available")
    else:
        print(f"✗ {col} - not found")
# look for the similar column names if some are missing
print(f"\nAll available columns:")
for col in df.columns:
    print(f"  {col}")

# update the feature list based on whats available
if 'neighbourhood_group' not in df.columns and 'neighbourhood_group_cleansed' in df.columns:
    available_features.append('neighbourhood_group_cleansed')
    print("Using 'neighbourhood_group_cleansed' instead of 'neighbourhood_group'")

if 'review_scores_rating' not in df.columns:
    # Look for other review score columns
    review_cols = [col for col in df.columns if 'review_score' in col.lower()]
    if review_cols:
        available_features.append(review_cols[0])
        print(f"Using '{review_cols[0]}' for review scores")

print(f"\nFinal feature list: {available_features}")
      
      
      
X = df[available_features].copy()
y = df['price'].copy()

print(f"\nFeature matrix shape: {X.shape}")
print(f"Target vector shape: {y.shape}")

      
print(f"\nHandling missing values...")
print("Missing values before cleaning:")
print(X.isnull().sum())

# fills in the missing values
for col in X.columns:
    if X[col].dtype in ['object']:  # categorical
        X[col] = X[col].fillna('Unknown')
    else:  # numerical
        X[col] = X[col].fillna(0)

print("Missing values after cleaning:")
print(X.isnull().sum())

      
print(f"\nEncoding categorical variables...")
categorical_columns = X.select_dtypes(include=['object']).columns
print(f"Categorical columns: {list(categorical_columns)}")

# convert boolean columns to 0/1
for col in X.columns:
    if X[col].dtype == 'bool':
        X[col] = X[col].astype(int)
        print(f"Converted {col} to 0/1")

# one-hot encode remaining categorical columns
X_encoded = pd.get_dummies(X, columns=categorical_columns, drop_first=True)
print(f"Final feature matrix shape after encoding: {X_encoded.shape}")

print("Final feature columns:")
for col in X_encoded.columns:
    print(f"  {col}")

# model training and evaluation

print("STEP 2: MODEL TRAINING AND EVALUATION")


# split the data
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print(f"Training set size: {X_train.shape[0]} samples")
print(f"Test set size: {X_test.shape[0]} samples")

# training the Random Forest model with default parameters
print(f"\nTraining Random Forest model...")
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

# evaluate the model
mae = mean_absolute_error(y_test, y_pred)
print(f"\nBaseline Model Performance:")
print(f"Mean Absolute Error: ${mae:.2f}")

from sklearn.metrics import mean_squared_error, r2_score
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"Root Mean Squared Error: ${rmse:.2f}")
print(f"R-squared Score: {r2:.3f}")

# feature importance
print(f"\nTop 10 Most Important Features:")
feature_importance = pd.DataFrame({
    'feature': X_encoded.columns,
    'importance': rf_model.feature_importances_
}).sort_values('importance', ascending=False)

print(feature_importance.head(10))


print("STEP 3: MODEL IMPROVEMENT")



print("Testing different n_estimators values...")
n_estimators_options = [50, 100, 200]
best_mae = float('inf')
best_n_estimators = None
results = []

for n_est in n_estimators_options:
    print(f"\nTesting n_estimators = {n_est}")
    model = RandomForestRegressor(n_estimators=n_est, random_state=42)
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    mae_score = mean_absolute_error(y_test, pred)
    
    results.append({
        'n_estimators': n_est,
        'mae': mae_score
    })
    
    print(f"MAE: ${mae_score:.2f}")
    
    if mae_score < best_mae:
        best_mae = mae_score
        best_n_estimators = n_est
        best_model = model

print(f"\nBest n_estimators: {best_n_estimators}")
print(f"Best MAE: ${best_mae:.2f}")

# training the final model with the best estimators
print(f"\nTraining final model with n_estimators = {best_n_estimators}")
final_model = RandomForestRegressor(n_estimators=best_n_estimators, random_state=42)
final_model.fit(X_train, y_train)


final_predictions = final_model.predict(X_test)
final_mae = mean_absolute_error(y_test, final_predictions)
final_rmse = np.sqrt(mean_squared_error(y_test, final_predictions))
final_r2 = r2_score(y_test, final_predictions)


print("FINAL MODEL RESULTS")
print(f"Mean Absolute Error: ${final_mae:.2f}")
print(f"Root Mean Squared Error: ${final_rmse:.2f}")
print(f"R-squared Score: {final_r2:.3f}")


# Visualizations
      
print("STEP 4: VISUALIZATIONS AND ANALYSIS")



fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Airbnb Price Prediction Model Results', fontsize=16)

# actual vs. predicted
axes[0,0].scatter(y_test, final_predictions, alpha=0.5)
axes[0,0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[0,0].set_xlabel('Actual Price ($)')
axes[0,0].set_ylabel('Predicted Price ($)')
axes[0,0].set_title('Actual vs Predicted Prices')

# residuals plot
residuals = y_test - final_predictions
axes[0,1].scatter(final_predictions, residuals, alpha=0.5)
axes[0,1].axhline(y=0, color='r', linestyle='--')
axes[0,1].set_xlabel('Predicted Price ($)')
axes[0,1].set_ylabel('Residuals ($)')
axes[0,1].set_title('Residuals Plot')

# features importance
top_features = feature_importance.head(8)
axes[1,0].barh(range(len(top_features)), top_features['importance'])
axes[1,0].set_yticks(range(len(top_features)))
axes[1,0].set_yticklabels(top_features['feature'])
axes[1,0].set_xlabel('Importance')
axes[1,0].set_title('Top 8 Feature Importances')

# model comparison
comparison_df = pd.DataFrame(results)
axes[1,1].bar(comparison_df['n_estimators'].astype(str), comparison_df['mae'])
axes[1,1].set_xlabel('Number of Estimators')
axes[1,1].set_ylabel('Mean Absolute Error ($)')
axes[1,1].set_title('Model Performance Comparison')

plt.tight_layout()
plt.show()
      
```
